# Exercices RL — CartPole-v1

## Étape 1 — Découverte de l'environnement

Objectif : créer l'environnement et explorer ses deux espaces fondamentaux :
- **l'espace d'observation** : ce que l'agent *voit*
- **l'espace d'action** : ce que l'agent *peut faire*

```python
import gymnasium as gym

env = gym.make("CartPole-v1")
env
```

### 1. Espace d'observation

```python
obs_space = env.observation_space

print("Espace d'observation :", obs_space)
print("Type               :", type(obs_space).__name__)
print("Forme (shape)      :", obs_space.shape)
print("dtype              :", obs_space.dtype)
print()

noms = ["Position du chariot", "Vitesse du chariot", "Angle du bâton (rad)", "Vitesse angulaire"]
for nom, low, high in zip(noms, obs_space.low, obs_space.high):
    print(f"{nom:<22} : [{low:>8.3f}, {high:>8.3f}]")
```

### 2. Espace d'action

```python
act_space = env.action_space

print("Espace d'action :", act_space)
print("Type            :", type(act_space).__name__)
print("Nombre d'actions:", act_space.n)
print()
print("0 -> pousser le chariot vers la GAUCHE")
print("1 -> pousser le chariot vers la DROITE")
```

### 3. Exemples aléatoires avec `.sample()`

```python
obs_space.seed(0)
act_space.seed(0)

print("Observations aléatoires :")
for _ in range(3):
    print("  ", obs_space.sample())

print("\nActions aléatoires :")
print("  ", [int(act_space.sample()) for _ in range(10)])
```

### 4. Bonus — une vraie observation (via `reset`)

`.sample()` tire n'importe quel point de l'espace. La vraie observation de départ, elle, vient de `env.reset()`.

```python
obs, info = env.reset(seed=42)
print("Observation initiale :", obs)
print("Dans l'espace ?      :", obs_space.contains(obs))
```

## Étape 2 — Boucle d'interaction avec un agent aléatoire

L'agent choisit ses actions au hasard pendant 10 épisodes. On affiche la récompense totale de chaque épisode.

```python
N_EPISODES = 10

env.action_space.seed(0)
rewards = []

for episode in range(N_EPISODES):
    obs, info = env.reset(seed=episode)
    total_reward = 0.0
    terminated = truncated = False

    while not (terminated or truncated):
        action = env.action_space.sample()
        obs, reward, terminated, truncated, info = env.step(action)
        total_reward += reward

    rewards.append(total_reward)
    fin = "terminated (bâton tombé / chariot sorti)" if terminated else "truncated (limite de 500 pas)"
    print(f"Épisode {episode + 1:>2} : récompense totale = {total_reward:>5.1f}  | fin : {fin}")

print(f"\nMoyenne sur {N_EPISODES} épisodes : {sum(rewards) / len(rewards):.1f}  (maximum possible : 500)")
```

```python
env.close()
```
